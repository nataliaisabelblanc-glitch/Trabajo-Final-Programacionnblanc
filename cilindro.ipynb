```python
import math
from scipy.optimize import minimize

n = 10
c1 = 2
c2 = 0.5

def funcion_objetivo(nUpla):
    h, r = nUpla # Extraer h y r de la lista nUpla
    return -n * math.pi * r**2 * h + c1 * math.pi * r**2 + c2 * math.pi * r**2 * h * (math.pi * r**2 + 2 * math.pi * r * h)

def restriccion1(nUpla):
    h, r = nUpla # Extraer h y r de la lista nUpla
    return h

def restriccion2(nUpla):
    h, r = nUpla # Extraer h y r de la lista nUpla
    return r

for i in range(10):
    conjetura = [i / 10, i / 10]
    solucion = minimize(funcion_objetivo, conjetura,
                        constraints=[{'type': 'ineq', 'fun': restriccion1},
                                     {'type': 'ineq', 'fun': restriccion2}])

    print(f"Para conjetura [{i/10},{i/10}]")
    print("success:", solucion.success)
    print("[h,r]=", solucion.x)
    print("Maximo:", -float(solucion.fun))
    print("===")
```
